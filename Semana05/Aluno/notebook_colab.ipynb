## 1. NumPy: arrays e operações vetorizadas

> 🔗 **Por que começar por aqui?** O Pandas, que você vai usar na sequência, é construído em cima do NumPy — cada coluna de um DataFrame é, por dentro, um array NumPy. Entender o NumPy primeiro deixa mais claro o que o Pandas está fazendo por trás dos panos.

### 🔹 Exemplo 1 — Array x lista

📖 **Antes do código: o que é `import numpy as np`?** Você já viu `import csv`, `import json`, `from openpyxl import load_workbook` — sempre importando pelo nome original. Agora aparece uma peça nova: `as np`. `as` dá um **apelido** pro que foi importado — depois de `import numpy as np`, você escreve `np.alguma_coisa` em vez de `numpy.alguma_coisa` inteiro, toda vez que for usar a biblioteca. `np` é o apelido praticamente universal do NumPy (todo tutorial, toda documentação usa esse mesmo apelido) — você vai ver `import pandas as pd` daqui a pouco, seguindo a mesma ideia.

`np.array(lista)` pega uma lista comum, que você já conhece, e devolve um **array NumPy** — por fora parece uma lista, mas se comporta de um jeito diferente nas contas, como você vai ver já na célula abaixo.

💡 **Este exemplo ainda não usa nenhum arquivo** — é só pra você comparar `list` (que você já conhece) com `array` (novo). Na sequência, você aplica isso aos dados reais da papelaria.

📦 **Antes do código: instalando o `numpy`.** Diferente de `csv`, `json`, `re` e `datetime` (que já vêm dentro do Python), o `numpy` é uma biblioteca **de terceiros** — feita por outras pessoas, não pelo Python.
```python
import numpy as np

precos_lista = [12.5, 2.0, 3.5]
precos_array = np.array([12.5, 2.0, 3.5])

print(type(precos_lista))
print(type(precos_array))
print(precos_array * 2)   # multiplica cada elemento por 2, de uma vez
```

💡 **O que aconteceu em cada linha:** `type(precos_lista)` mostra `<class 'list'>`, como você já esperava. `type(precos_array)` mostra `<class 'numpy.ndarray'>` — um tipo novo. O nome `ndarray` significa **"n-dimensional array"** ("array de N dimensões"): o "n" é só um espaço reservado pra "quantas dimensões esse array tiver". Os arrays de hoje são todos **1D** (uma sequência simples de números, igual a uma lista) — mas o NumPy também lida com arrays **2D** (uma tabela, linhas x colunas — é exatamente o que vira uma coluna de DataFrame mais adiante), 3D, e assim por diante, todos usando essa mesma classe genérica `ndarray`. Por fim, `precos_array * 2` devolve `[25. 4. 7.]` — cada valor foi multiplicado por 2, de forma independente.

> ⚠️ **Veja como é um erro real do Python.** Multiplicar uma lista comum por 2, `precos_lista * 2`, não dá erro — mas também não faz o que parece: em vez de dobrar cada preço, o Python **repete a lista duas vezes seguidas**, `[12.5, 2.0, 3.5, 12.5, 2.0, 3.5]`. É um dos bugs mais traiçoeiros de quem vem de lista pra array: não trava, só faz a coisa errada silenciosamente. Teste na célula abaixo.

```python
precos_lista = [12.5, 2.0, 3.5]
print(precos_lista * 2)   # repete a lista, não multiplica cada valor
```

### 🔹 Exemplo extra — Somar um número: lista dá erro, array não

📖 **Antes do código:** você acabou de ver que `lista * 2` não trava, só repete a lista (o bug silencioso). Somar um número a uma lista é diferente: `lista + 1` **trava de verdade**, com `TypeError` — o Python não sabe como somar um número a uma lista, e nem tenta adivinhar. Com array, `array + 1` soma 1 em **cada posição** de uma vez — isso é broadcasting, que você vai ver com mais detalhe daqui a pouco nesta mesma seção.

```python
import numpy as np
try:
    lista = [1, 2, 3, 4, 5]
    print(lista + 1)
except TypeError as e:
    print(f"Erro: {e}")

array = np.array([1, 2, 3, 4, 5])
print(array + 1)
```

💡 `lista * 2` (multiplicação) e `lista + 1` (soma) se comportam **diferente** um do outro: multiplicar por um número repete a lista inteira; somar um número nem tenta, e trava com erro. Duas armadilhas diferentes — o que importa é lembrar que **nenhuma das duas operações faz, numa lista comum, o que parece que vai fazer**.

### 🔹 Exemplo extra — Nem toda lista vira um array "limpo"

📖 **Antes do código:** uma lista pode misturar tipos diferentes na mesma variável — número, texto, o que for. Um array NumPy **não pode**: todos os elementos precisam ser do mesmo tipo. Se você passar uma lista com tipos misturados pro `np.array(...)`, o NumPy não recusa — ele **converte tudo pro tipo mais genérico que consiga representar todos os valores** (aqui, texto, já que texto consegue representar qualquer coisa).

```python
import numpy as np
lista_misturada = [1, "dois", 3.0]
array_misturado = np.array(lista_misturada)

print(array_misturado)
print(array_misturado.dtype)
```

💡 **`dtype`** ("data type") é um atributo que todo array carrega, mostrando o tipo dos elementos guardados nele — aqui, `<U32` significa "texto (até 32 caracteres)". Repare que `1` (inteiro) e `3.0` (decimal) viraram texto também, só pra combinar com `'dois'`. É por isso que, no Exemplo 2 mais adiante, ao ler os dados reais do CSV, sempre convertemos com `float(...)`/`int(...)` antes de montar o array: um único valor de texto sobrando no meio faria o array inteiro virar texto, e a multiplicação de preços quebraria.

### 🔹 Exemplo extra — Por que isso importa: velocidade

📖 **Antes do código:** com poucos elementos, a diferença de velocidade entre lista e array é imperceptível. Mas datasets reais têm milhares (ou milhões) de linhas — e aí a diferença fica bem visível. `time.time()` marca "que horas são agora" em segundos; a diferença entre duas marcações é quanto tempo o código levou pra rodar.

```python
import numpy as np
import time

numeros_lista = list(range(1, 10_000_001))
numeros_array = np.array(numeros_lista)

inicio = time.time()
soma_lista = sum(numeros_lista)
print(f"Lista: {time.time() - inicio:.4f} segundos")

inicio = time.time()
soma_array = np.sum(numeros_array)
print(f"Array: {time.time() - inicio:.4f} segundos")
```

💡 Rode essa célula de novo — o tempo exato varia a cada execução (depende do que mais o computador está fazendo no momento), mas o array vai ser sempre bem mais rápido, geralmente várias vezes mais rápido que a lista, pra somar os mesmos 10 milhões de números. É por isso que, quando o assunto é dado numérico em volume real, NumPy — e o Pandas, que é construído em cima dele — é a ferramenta padrão do mercado: não só por ser mais prático de escrever, mas por ser muito mais rápido de rodar.

### 🔹 Exemplo extra — Fatiando um array (slice)

📖 **Antes do código: o que é fatiamento (`slice`)?** `array[início:fim]` devolve um pedaço do array, do índice `início` até o índice `fim` — mas **sem incluir** o `fim`. Num array de 5 posições (índices 0 a 4), `array[1:4]` devolve os índices 1, 2 e 3 — três valores, não quatro. Isso não é exclusividade do NumPy: listas comuns, tuplas e até texto (`string`) já funcionam com essa mesma sintaxe `[início:fim]` em Python — só que ninguém tinha precisado ainda. Dois atalhos úteis: `array[:3]` (sem número antes dos dois-pontos) significa "do começo até o índice 3, exclusive"; `array[3:]` (sem número depois) significa "do índice 3 até o final".

```python
import numpy as np
precos_array = np.array([12.5, 2.0, 3.5, 6.0, 150.0])

print(precos_array[1:4])   # índices 1, 2 e 3 -- NÃO inclui o índice 4
print(precos_array[:3])    # do início até o índice 3 (exclusive) -- os 3 primeiros
print(precos_array[3:])    # do índice 3 até o final
```

💡 **Por que o `fim` fica de fora?** Duas razões práticas: `fim - início` já te dá direto a quantidade de elementos (`[1:4]` → `4 - 1 = 3` elementos); e fatias vizinhas se encaixam sem sobrepor nem pular nada — `array[:3]` + `array[3:]` cobre o array inteiro, cada índice aparecendo uma única vez.

> ⚠️ **Erro comum:** pedir mais índices do que o array tem, tipo `precos_array[3:10]`, **não trava** — o Python devolve o que existir a partir do índice 3, sem erro nenhum e sem avisar que você pediu mais do que existia. Repare a diferença com o erro de broadcasting que você vai ver daqui a pouco: lá, tamanhos incompatíveis multiplicando dão erro; aqui, um intervalo maior que o array simplesmente devolve o que existe. Teste na célula abaixo.

```python
import numpy as np

precos_array = np.array([12.5, 2.0, 3.5, 6.0, 150.0])
print(precos_array[3:10])   # não existe índice 10, mas não dá erro nenhum
```

Esse é o primeiro registro do arquivo inteiro (Caderno, Janeiro) — repare nas chaves do dicionário: `produto`, `categoria`, `mes`, `preco_unitario`, `quantidade_vendida`. São exatamente essas 5 chaves que você vai usar pra filtrar e calcular daqui pra frente — nenhum nome de coluna "cai do céu" no código que vem a seguir.

Agora que você já sabe que existe uma coluna `mes`, filtrar por ela faz sentido. Vamos buscar só os registros de Janeiro e ver como ficam.

```python
import csv

with open("dataset/vendas_papelaria.csv", encoding="utf-8") as arquivo:
    vendas_jan = [linha for linha in csv.DictReader(arquivo) if linha["mes"] == "Jan"]

for registro in vendas_jan:
    print(registro)
```

Repare nos 6 registros de Janeiro, um por linha — cada um ainda é um dicionário com valores em texto (`'12.5'`, não `12.5`). Agora que você já viu como são esses dados, vamos transformar as colunas `preco_unitario` e `quantidade_vendida` em arrays NumPy pra calcular a receita.

```python
import csv
import numpy as np

with open("dataset/vendas_papelaria.csv", encoding="utf-8") as arquivo:
    vendas_jan = [linha for linha in csv.DictReader(arquivo) if linha["mes"] == "Jan"]

precos = np.array([float(linha["preco_unitario"]) for linha in vendas_jan])
quantidades = np.array([int(linha["quantidade_vendida"]) for linha in vendas_jan])

receita = precos * quantidades   # multiplicação elemento a elemento, sem for!
print(receita)
```

💡 **O que aconteceu em cada linha:** `[linha for linha in csv.DictReader(arquivo) if linha["mes"] == "Jan"]` é a list comprehension que você já viu na Semana 04, filtrando só as linhas de Janeiro. Cada valor lido de um CSV chega como texto — por isso `float(...)` e `int(...)` convertem `"12.5"` e `"30"` para número antes de entrar no array; sem essa conversão, `np.array` criaria um array de texto, e `*` entre textos não multiplica, dá erro. `receita = precos * quantidades` multiplica **posição a posição**: o primeiro preço pelo primeiro quantidade, o segundo pelo segundo, e assim por diante — os dois arrays precisam ter o mesmo tamanho pra isso funcionar.

### 🔹 Exemplo extra — Broadcasting e funções estatísticas

📖 **Antes do código:** multiplicar um array inteiro por um único número, como `precos * 1.10`, aplica essa conta em **cada posição** do array de uma vez — isso se chama **broadcasting**. `.sum()`, `.mean()` e `.max()` são métodos que já vêm prontos em qualquer array NumPy: somam, tiram a média e acham o maior valor, sem precisar escrever um `for`.

```python
import csv
import numpy as np

with open("dataset/vendas_papelaria.csv", encoding="utf-8") as arquivo:
    vendas_jan = [linha for linha in csv.DictReader(arquivo) if linha["mes"] == "Jan"]

precos = np.array([float(linha["preco_unitario"]) for linha in vendas_jan])
quantidades = np.array([int(linha["quantidade_vendida"]) for linha in vendas_jan])
receita = precos * quantidades

precos_com_aumento = precos * 1.10   # aumenta 10% em todos de uma vez, sem for
print(precos_com_aumento)

print(f"Receita total de janeiro: {receita.sum()}")
print(f"Preço médio: {precos.mean():.2f}")
print(f"Maior preço: {precos.max()}")
```

💡 **Isso é broadcasting:** `precos * 1.10` multiplica um array inteiro por um único número — o NumPy "estica" o `1.10` pra combinar com cada posição do array, sem você precisar escrever isso manualmente. `.sum()`, `.mean()` e `.max()` são métodos que já vêm prontos no array, poupando um `for` pra somar, calcular média ou achar o maior valor.

> ⚠️ **Quando o broadcasting NÃO funciona:** arrays de tamanhos incompatíveis geram um erro real, não um resultado "estranho" — veja na célula abaixo.

```python
import numpy as np
try:
    a = np.array([1, 2, 3])
    b = np.array([1, 2])
    print(a * b)
except ValueError as e:
    print(f"Erro: {e}")
```

### 🔹 Exemplo extra — Series x DataFrame

📖 **Antes do código:** `vendas["preco_unitario"]`, com **um par** de colchetes, devolve uma **Series** — uma coluna solta, sem a estrutura de tabela. `vendas[["preco_unitario"]]`, com **dois pares** (uma lista com um nome dentro), devolve um **DataFrame** de uma coluna só, mantendo a tabela. `type(...)` na célula abaixo confirma essa diferença.

## 3. Explorando um DataFrame: describe, filtros e groupby

> 🔗 **Por que isso agora?** Todo número que `describe()` calcula a seguir — média, desvio, mínimo, máximo — é o NumPy rodando por baixo em cada coluna, a mesma "conta em bloco, sem `for`" da Seção 1. A diferença é que agora você não escreve `np.mean(coluna)` uma coluna de cada vez: o Pandas já sabe quais colunas são numéricas e aplica isso em todas de uma vez, com um comando só.

### 🔹 Exemplo 1 — describe()

📖 **Antes do código:** `describe()` é um método que já vem pronto em qualquer DataFrame — ele calcula, de uma vez, um resumo estatístico de cada coluna **numérica**: quantos valores existem (`count`), a média (`mean`), o quanto os valores variam em torno da média (`std`), o menor e o maior valor (`min`/`max`), e 3 pontos intermediários chamados quartis (`25%`, `50%`, `75%` — por exemplo, `25%` é o valor abaixo do qual ficam 25% dos dados). É o primeiro comando que um analista roda ao abrir qualquer dataset novo, antes de qualquer gráfico.

```python
import pandas as pd
from IPython.display import display

vendas = pd.read_csv("dataset/vendas_papelaria.csv")
display(vendas.describe())
```

💡 **Lendo o resultado:** repare que `preco_unitario` varia bem mais que `quantidade_vendida`, proporcionalmente — o `std` (53.8) é quase 1,5x a própria média (34.8), puxado pela Mochila, que custa muito mais que os outros produtos. Isso já é um sinal de que vale a pena olhar os dados por categoria, não só o dataset inteiro — é exatamente o que o `groupby` faz mais adiante.

### Selecionando por posição ou por rótulo: `.loc` e `.iloc`

`.loc` usa rótulos (nomes), `.iloc` usa posições (números). Os dois fazem a mesma coisa de jeitos diferentes — use `.loc` quando você sabe o nome da coluna, e `.iloc` quando só sabe a posição. Antes de acessar uma posição específica, vale ver a tabela de novo — "linha 0" só faz sentido depois que você viu como ela é.

```python
import pandas as pd
from IPython.display import display

vendas = pd.read_csv("dataset/vendas_papelaria.csv")
display(vendas.head())
```

A linha 0 (a primeira) é o Caderno de Janeiro — é essa linha que `.loc[0, "produto"]` e `.iloc[0, 0]` vão acessar na célula seguinte. Sem ver a tabela primeiro, "linha 0, coluna produto" fica abstrato; vendo, fica óbvio que o resultado tem que ser `"Caderno"`.

### 🔹 Exemplo 2 — Filtro e groupby

📖 **Antes do código:** `vendas[vendas["quantidade_vendida"] > 40]` é um **filtro** — dentro dos colchetes de fora, `vendas["quantidade_vendida"] > 40` compara cada linha e devolve `True`/`False`; só as linhas `True` aparecem no resultado. `vendas.groupby("categoria")["preco_unitario"].mean()` **agrupa** as linhas que têm o mesmo valor de `categoria`, e calcula a média de `preco_unitario` **dentro de cada grupo** — o passo a passo completo de como isso funciona vem na caixa logo depois do código.

📌 Repare que esta célula reimporta `pandas` e reabre `vendas_papelaria.csv` de novo, mesmo você já tendo feito isso antes — assim ela funciona sozinha, mesmo que você rode só ela depois de reabrir o notebook.

```python
import pandas as pd
from IPython.display import display

vendas = pd.read_csv("dataset/vendas_papelaria.csv")

# Filtro: só produtos vendidos mais de 40 vezes no mês
mais_vendidos = vendas[vendas["quantidade_vendida"] > 40]
display(mais_vendidos)

# groupby: preço médio por categoria
display(vendas.groupby("categoria")["preco_unitario"].mean())
```

### Como o groupby pensa por dentro: Dividir → Aplicar → Combinar

1. **Dividir (split):** o Pandas separa as linhas em grupos, um para cada valor de `categoria` (`Papelaria`, `Acessorios`).
2. **Aplicar (apply):** dentro de cada grupo separado, calcula a função pedida (ex.: `mean()` dos preços apenas daquele grupo).
3. **Combinar (combine):** junta o resultado de cada grupo numa única Series de resposta.

É exatamente isso que a Tabela Dinâmica do Excel faz por trás dos panos — o `groupby` só está explicitando os 3 passos acima. Repare que `Acessorios` (Mochila e Estojo) tem preço médio bem mais alto que `Papelaria` (Caderno, Lápis, Caneta, Marca-texto) — faz sentido, já que mochila e estojo custam mais que material de papelaria simples.

### Ordenando o resultado: `sort_values()`

`sort_values()` reordena o resultado. O parâmetro `ascending` controla a direção: `ascending=False` quer dizer "não crescente", ou seja, do **maior pro menor**; `ascending=True` (o padrão, se você omitir) ordena do **menor pro maior**. Útil pra responder "qual categoria tem o maior preço médio?" sem precisar procurar manualmente o maior número na tabela.

📌 Repare que esta célula reimporta `pandas` e reabre `vendas_papelaria.csv` de novo, mesmo você já tendo feito isso antes — assim ela funciona sozinha, mesmo que você rode só ela depois de reabrir o notebook.

```python
import pandas as pd
from IPython.display import display

vendas = pd.read_csv("dataset/vendas_papelaria.csv")

preco_medio_por_categoria = vendas.groupby("categoria")["preco_unitario"].mean()
display(preco_medio_por_categoria.sort_values(ascending=False))
```

## 4. Combinando tudo: receita por categoria

📖 **Antes do código:** `vendas["receita"] = ...` atribui um resultado a uma coluna chamada `receita` — como essa coluna ainda não existe no DataFrame, o Pandas simplesmente **cria ela**, sem precisar de nenhum comando especial de "adicionar coluna". A conta usada, `preco_unitario * quantidade_vendida`, é uma operação vetorizada do próprio Pandas (que por trás usa NumPy, exatamente como você viu na Seção 1) — roda linha por linha, de uma vez, sem `for`.

📌 Repare que esta célula reimporta `pandas` e reabre `vendas_papelaria.csv` de novo, mesmo você já tendo feito isso antes — assim ela funciona sozinha, mesmo que você rode só ela depois de reabrir o notebook.

```python
import pandas as pd
from IPython.display import display

vendas = pd.read_csv("dataset/vendas_papelaria.csv")

vendas["receita"] = vendas["preco_unitario"] * vendas["quantidade_vendida"]
display(vendas.head())

receita_por_categoria = vendas.groupby("categoria")["receita"].sum()
display(receita_por_categoria.sort_values(ascending=False))
```

💡 **Lendo o resultado:** `Acessorios` (Mochila + Estojo) gera mais receita total que `Papelaria`, mesmo vendendo muito menos unidades — o preço unitário mais alto compensa o volume menor. É o mesmo padrão que você já viu no `describe()`: números absolutos (receita total) e números relativos (preço médio) contam histórias diferentes sobre o mesmo dataset.

### 🔹 Exemplo extra — Por dentro, uma coluna do Pandas é um array NumPy

📖 **Antes do código:** `.to_numpy()` converte uma coluna do Pandas (uma Series) num array NumPy puro — sem rótulos, só os valores. Isso prova concretamente o que a Seção 1 já antecipou: por baixo dos panos, toda coluna numérica de um DataFrame **já é** um array NumPy, só que com um índice de linha e um nome de coluna grudados nela por fora.

```python
import pandas as pd

vendas = pd.read_csv("dataset/vendas_papelaria.csv")
vendas["receita"] = vendas["preco_unitario"] * vendas["quantidade_vendida"]

receita_array = vendas["receita"].to_numpy()
print(receita_array)
print(type(receita_array))
```

Repare que `receita_array` é só uma sequência de números — sem nome de coluna, sem índice de linha, `type()` confirma `numpy.ndarray`. Agora vamos comparar a média calculada de dois jeitos diferentes em cima desse mesmo array.

```python
import pandas as pd
import numpy as np

vendas = pd.read_csv("dataset/vendas_papelaria.csv")
vendas["receita"] = vendas["preco_unitario"] * vendas["quantidade_vendida"]

receita_array = vendas["receita"].to_numpy()

print(f"Média via NumPy:  {np.mean(receita_array):.2f}")
print(f"Média via Pandas: {vendas['receita'].mean():.2f}")
```

💡 **Mesma conta, dois caminhos, resultado idêntico:** `np.mean(receita_array)` calcula a média chamando o NumPy diretamente sobre o array puro; `vendas["receita"].mean()` calcula a mesma média usando o método do Pandas. Os dois dão `406.04` — não é coincidência: por dentro, o método `.mean()` do Pandas **chama o NumPy** pra fazer a conta. Você não precisa escolher um ou outro; o Pandas existe justamente pra te dar a comodidade de método (`.mean()`) em cima do que o NumPy já faz bem (calcular rápido em cima de um array).
